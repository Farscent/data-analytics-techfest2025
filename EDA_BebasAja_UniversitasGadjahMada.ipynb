**Latar Belakang**

Transfer ke Daerah dan Dana Desa (TKDD) adalah salah satu alat penting dalam kebijakan fiskal Indonesia yang bertujuan mendorong desentralisasi dan otonomi daerah. Alokasi TKDD digunakan untuk membiayai berbagai kebutuhan di daerah dan desa, antara lain Dana Alokasi Umum (DAU), Dana Bagi Hasil (DBH), Dana Alokasi Khusus (DAK), Dana Otonomi Khusus, serta Dana Desa sebesar 70 triliun rupiah yang didistribusikan ke hampir 75 ribu desa di seluruh Indonesia [(Direktorat Jenderal Perimbangan Keuangan, 2022)](https://djpk.kemenkeu.go.id/?p=27451).

Peningkatan alokasi TKDD dilakukan untuk memperkuat kemampuan keuangan daerah dalam memberikan pelayanan publik yang lebih baik, mempercepat pembangunan infrastruktur, serta meningkatkan taraf hidup masyarakat di tingkat daerah. Dalam mengukur sejauh mana pembangunan tersebut berhasil, salah satu indikator yang sering digunakan adalah Indeks Pembangunan Manusia (IPM). IPM di suatu daerah adalah gambaran mengenai sejauh mana kemajuan pembangunan di wilayah tersebut, terutama dalam hal pengembangan sumber daya manusia [(Fahrurrozi et al., 2023)](https://eprints.hamzanwadi.ac.id/5188/). Dalam penerapannya, IPM menjadi pedoman penting bagi pemerintah daerah dalam mengevaluasi hasil dari upaya pembangunan yang telah dilakukan. IPM mencerminkan kualitas pembangunan dari tiga aspek utama: kesehatan, pendidikan, dan standar hidup yang layak.

Di sisi lain, distribusi TKDD yang belum merata dan pengelolaan anggaran yang kurang maksimal bisa menjadi salah satu penyebab munculnya kesenjangan pembangunan antar daerah. Oleh karena itu, penting untuk melihat sejauh mana distribusi TKDD pada tahun 2023 benar-benar berdampak terhadap peningkatan IPM di berbagai wilayah Indonesia. Melalui analisis ini, diharapkan muncul gambaran yang lebih jelas tentang seberapa efektif kebijakan TKDD dalam mendukung pembangunan manusia, sekaligus mengidentifikasi kendala yang mungkin masih perlu dibenahi agar tujuan pemerataan pembangunan bisa tercapai.

Dengan memahami kaitan antara distribusi TKDD dan IPM, pemerintah dapat menyusun strategi pengalokasian anggaran yang lebih tepat sasaran. Selain itu, pemahaman ini juga bisa mendorong peningkatan transparansi dan akuntabilitas dalam penggunaan dana, serta memperkuat kerja sama antara pemerintah pusat dan daerah dalam upaya meningkatkan kualitas hidup masyarakat secara lebih merata.

Hal ini merujuk pada data alokasi TKDD tahun 2023 dan konteks peran TKDD dalam pembangunan daerah serta relevansinya terhadap IPM sebagai indikator pembangunan manusia di Indonesia.

## Mengubah tipe data objek ke numerik (integer atau float)
Bisa kita lihat bahwa dari dataset kita ada kolom-kolom yang berbentuk numerik, tetapi masih berbentuk objek. Kita harus mengubah kolom-kolom tersebut agar menjadi numerik (integer atau float) supaya dapat kita analisis. Berikut adalah kode untuk mengubah tipe data objek ke numerik.

```python
def clean_number(val):
    if pd.isna(val):
        return np.nan
    val = str(val)
    val = val.replace('Rp', '')
    val = val.replace('.', '')      # hapus titik ribuan
    val = val.replace(',', '.')     # ubah koma ke titik desimal
    val = val.strip()               # hapus spasi kiri-kanan
    try:
        return float(val)
    except:
        return np.nan

# Terapkan ke semua kolom kecuali 'Provinsi'
for col in df.columns:
    if col != "Provinsi":
        df[col] = df[col].apply(clean_number)
```

### Catatan mengenai karakteristik data numerik keseluruhan
Di bagian ini kami akan menganalisis data pencilan, data minimum, dan data maximum sebagai acuan kita di bagian **Pembahasan**. *Outlier* merupakan data yang memiliki nilai yang sangat berbeda jauh dibandingkan dengan sebagian besar data lainnya. Keberadaan *outlier* tidak hanya dapat menyebabkan distribusi data menjadi tidak normal, tetapi juga dapat memengaruhi validitas dalam penarikan kesimpulan atau pengambilan keputusan dalam suatu penelitian. Suatu data dapat  dikatakan sebagai data outlier jika nilai observasi lebih  kecil dari Q1−1.5∗IQR atau nilainya lebih besar dari Q3+1.5∗IQR [(Sihombing et al., 2022)](https://jurnaljesi.com/index.php/jurnaljesi/article/view/112).



```python
warnings.filterwarnings("ignore")

kolom_numerik = df.select_dtypes(include='number').columns.tolist()

def format_angka(x, _):
    if x >= 1e12:
        return f'{x/1e12:.1f} Triliun'
    elif x >= 1e9:
        return f'{x/1e9:.1f} Miliar'
    elif x >= 1e6:
        return f'{x/1e6:.1f} Juta'
    elif x >= 1e3:
        return f'{x/1e3:.1f} Ribu'
    else:
        return f'{x:,.0f}'

for col in kolom_numerik:
    plt.figure(figsize=(8, 1.5))
    sns.boxplot(x=df[col], color='skyblue')
    plt.title(f'Distribusi Nasional: {col}')
    plt.gca().xaxis.set_major_formatter(ticker.FuncFormatter(format_angka))
    plt.tight_layout()
    plt.show()
```

```python
kolom_numerik = df.select_dtypes(include='number').columns.tolist()

for col in kolom_numerik:
    col_name = col.lower().replace(' ', '_')

    print(f'\nKolom: {col}')
    print('-' * 50)

    min_value = df[col].min()
    max_value = df[col].max()

    min_rows = df[df[col] == min_value][['Provinsi', col]].copy()
    max_rows = df[df[col] == max_value][['Provinsi', col]].copy()

    print('Baris dengan nilai MIN:')
    display(min_rows)

    print('\nBaris dengan nilai MAX:')
    display(max_rows)

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_rows = df[(df[col] < lower_bound) | (df[col] > upper_bound)][['Provinsi', col]].copy()

    print(f'\nJumlah outlier: {len(outlier_rows)}')
    if not outlier_rows.empty:
        display(outlier_rows)

    df_note = pd.concat([max_rows, min_rows, outlier_rows], ignore_index=True)

    vars()[f'df_note_{col_name}'] = df_note
```

##  Mengatasi kolom yang mengandung nilai yang hilang
Kami akan membagi menjadi 2 dataframe untuk analisis kami, yaitu:
1. **df**: Dataframe tanpa melakukan imputasi untuk nilai yang hilang (dibiarkan saja).
2. **df_imputed**: Dataframe dengan melakukan imputasi untuk nilai yang hilang dengan membuat *grouping* dengan mengelompokkan menjadi 6 pulau besar di Indonesia, yaitu Sumatera, Jawa, Kalimantan, Sulawesi, Papua, dan Bali & Nusa Tenggara. Alasan menggunakan metode imputasi ini telah dijabarkan di subbab di bawah ini **(subbab: Alasan menggunakan metode imputasi)**.

```python
df_imputed = df.copy()

kolom_numerik = df_imputed.select_dtypes(include='number').columns

for col in kolom_numerik:
    df_imputed[col] = df_imputed.groupby('Pulau')[col].transform(lambda x: x.fillna(x.median()))
```

### Uji hipotesis dengan korelasi Pearson
Uji korelasi Pearson, atau dikenal juga sebagai Pearson Product Moment Correlation, adalah metode statistik yang digunakan untuk mengukur kekuatan dan arah hubungan linier antara dua variabel numerik (interval atau rasio) [(Hulu & Sinaga, 2019)](https://www.scribd.com/document/771691897/ANALISIS-KORELASI). Nilai koefisien korelasi Pearson (r) berkisar antara -1 hingga 1:

r = 1: hubungan positif sempurna,

r = -1: hubungan negatif sempurna,

r = 0: tidak ada hubungan linier.

- yi = nilai y ke-i

---



Pertama, kita harus memahami hipotesis yang akan di jawab oleh uji korelasi pearson. Hipotesis uji korelasi pearson sebagai berikut:

H0 : Tidak terdapat hubungan yang signifikan antara Realisai TKDD dan IPM (P-value > 0.05)

H1 : Terdapat hubungan yang signifikan antara Realisai TKDD dan IPM (P-value < 0.05)


```python
plt.figure(figsize=(8, 5))
sns.regplot(x='Realisasi TKDD', y='IPM', data=df_imputed, color='skyblue')

corr, p_value = stats.pearsonr(df_imputed['Realisasi TKDD'], df_imputed['IPM'])
plt.title(f'Relasi antara Realisasi TKDD dan IPM\nKoefisien Korelasi: {corr:.2f}, p-value: {p_value:.4f}')
plt.xlabel('Realisasi TKDD (Triliun Rupiah)')
plt.ylabel('Indeks Pembangunan Manusia (IPM)')

plt.gca().xaxis.set_major_formatter(ticker.FuncFormatter(lambda x, _: f'{x/1e12:.1f} T'))

plt.tight_layout()
plt.show()
```

Berdasarkan hasil yang kita dapatkan bahwa korelasi pearson yang dihasilkan oleh Realisasi TKDD dan IPM menghasilkan **p-value 0**.1578 yang artinya kita menolak H1 sehingga **tidak terdapat hubungan yang signifikan antara Realisai TKDD dan IPM**. Ditambah lagi **koefisien korelasi** di antara keduanya menunjukkan angka **0.23** yang artinya hubungan antara keduanya **lemah**.

**Insights:**
- Tidak adanya hubungan signifikan secara statistik antara Realisasi TKDD dan IPM menunjukkan bahwa tingginya anggaran yang direalisasikan oleh suatu provinsi belum tentu berdampak langsung terhadap peningkatan kualitas hidup masyarakat (IPM).

- Koefisien korelasi yang lemah (0.23) juga mengindikasikan bahwa meskipun ada sedikit kecenderungan positif, faktor-faktor lain kemungkinan jauh lebih berpengaruh dalam meningkatkan IPM, seperti efektivitas penggunaan dana, kualitas program pembangunan, kondisi geografis, pendidikan, dan kesehatan.

### Pengelompokan Provinsi Berdasarkan Realisasi TKDD dan IPM (*clustering*)

*Clustering* merupakan teknik analisis data yang digunakan untuk mengelompokkan provinsi berdasarkan kemiripan karakteristik, seperti Indeks Pembangunan Manusia (IPM) dan realisasi anggaran TKDD [(Rajsya, Purnawansyah, & Manga, 2024)](https://jurnal.fikom.umi.ac.id/index.php/LINIER/article/view/2539). Salah satu metode yang populer adalah K-Means karena efisien dan mudah diimplementasikan.

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

X = df_imputed[['IPM', 'Realisasi TKDD']]

inertia = []
K = range(1, 10)
for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(6, 4))
plt.plot(K, inertia, 'bo-')
plt.xlabel('Jumlah Cluster (k)')
plt.ylabel('Inertia')
plt.title('Metode Elbow untuk Menentukan k')
plt.grid(True)
plt.tight_layout()
plt.show()
```

Kami memutuskan untuk meng-*cluster* menjadi 3 kluster. Hal ini didasari oleh metode yang kami gunakan, yaitu *Elbow Method*. *Elbow Method* adalah teknik yang umum digunakan untuk menentukan jumlah klaster optimal dalam algoritma k-means clustering. Metode ini bekerja dengan cara memplot nilai Within-Cluster Sum of Squares (WCSS) terhadap jumlah klaster (K) yang berbeda. WCSS mengukur total jarak kuadrat antara setiap titik data dengan pusat klasternya, sehingga nilai WCSS yang lebih kecil menunjukkan klaster yang lebih rapat dan baik [(Rajsya, Purnawansyah, & Manga, 2024)](https://jurnal.fikom.umi.ac.id/index.php/LINIER/article/view/2539). Dengan *Elbow Method*, kami menemukan fakta bahwa kami menemukan titik *elbow* atau titik di mana inersia turun secara signifikan adalah ketika jumlah kluster sama dengan 3. Oleh karena itu, kami menggunakan k=3 sebagai jumlah kluster sebagai acuan analisis kami dalam mencari pola antara Realisasi TKDD dengan IPM.

```python
kmeans = KMeans(n_clusters=3, random_state=42)
df_imputed['Cluster'] = kmeans.fit_predict(X)

plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df_imputed,
    x='Realisasi TKDD',
    y='IPM',
    hue='Cluster',
    palette='Set2',
    s=100,
    edgecolor='black'
)

plt.title('Hasil Clustering Provinsi berdasarkan IPM dan Realisasi TKDD')
plt.xlabel('Realisasi TKDD (Rp)')
plt.ylabel('Indeks Pembangunan Manusia (IPM)')

plt.gca().xaxis.set_major_formatter(ticker.FuncFormatter(lambda x, _: f'{x/1e12:.1f} T'))

plt.legend(title='Cluster')
plt.tight_layout()
plt.show()
```
